```python
import zindi

client = zindi.Client()
client
```

## Browse competitions

No account needed. Try `status="all"` to practise on closed competitions.

```python
client.competitions(status="open", limit=10)
```

## Or explore with buttons

```python
zindi.explore(client)
```

## Pick one

Replace the id with one from the list above.

```python
COMPETITION = "Crop-Yield-Forecast-Challenge"

comp = client.competition(COMPETITION)
comp
```

```python
print(comp.rules[:1500])
```

## Join and download the data

Joining means accepting the competition rules. Downloads are skipped on re-runs when the files are already there.

```python
client.use(COMPETITION)
client.join()
paths = client.download(destination="data")
paths
```

## Build a submission

Here we simply submit the sample submission to check that everything works. Replace this with your model.

```python
import pandas as pd

sample = next(p for p in paths if "sample" in p.name.lower())
submission = pd.read_csv(sample)
submission.to_csv("submission.csv", index=False)
submission.head()
```

## Submit and wait for the score

```python
sub = client.submit("submission.csv", comment="sample submission from the quick start", wait=True)
sub
```

```python
print("Submissions left today:", client.limits().remaining_today)
client.submissions()
```

## Where do I stand?

```python
print("My rank:", client.rank())
client.leaderboard(limit=15)
```
